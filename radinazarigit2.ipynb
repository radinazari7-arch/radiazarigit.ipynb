```python
import pandas as pd

df = pd.read_csv('product_reviews_dataset.csv')

df
```

```python
df.isnull().sum()
```

```python
import re

def clean_text(text):
    text = re.sub(r'[^\w\s]' , ' ' , text)
    text = re.sub(r'[😊👍]', '', text)
    text = text.lower()

    return text 
```

```python
df['clean'] = df['review'].apply(clean_text)

df
```

```python
from sklearn.preprocessing import LabelEncoder

R = LabelEncoder()
y = R.fit_transform(df['sentiment'])
```

```python
x = df['clean']
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(
    x, y , train_size=0.8 , random_state= 42 , stratify=y
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

D = TfidfVectorizer(max_features=2000 , ngram_range=(1 , 2))
x_train = D.fit_transform(x_train)
x_test = D.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(x_train , y_train)
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score , confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns 


accuracy_score(y_test , y_pred)
```

```python
N = confusion_matrix(y_test , y_pred)
plt.figure(figsize=(6 , 4))
sns.heatmap(N ,  annot=True , fmt='d' , cmap='Blues')
```
